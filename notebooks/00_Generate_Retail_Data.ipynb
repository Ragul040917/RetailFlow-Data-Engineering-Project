```python

import pandas as pd
import random
import os
from datetime import datetime, timedelta

# Reproducible datasets
random.seed(42)

# Volume location
BASE_PATH = "/Volumes/dbacademy/retailflow/raw_data"

os.makedirs(BASE_PATH, exist_ok=True)

# -----------------------------
# 1. PRODUCTS DATASET
# -----------------------------

products = []

categories = ["Electronics", "Clothing", "Groceries", "Furniture", None]

for i in range(1, 51):
    products.append({
        "product_id": i,
        "product_name": f"Product_{i}",
        "category": random.choice(categories),
        "price": round(random.uniform(100, 5000), 2)
    })

products_df = pd.DataFrame(products)

# -----------------------------
# 2. STORES DATASET
# -----------------------------

cities = [
    "Chennai", "Bangalore", "Hyderabad", "Mumbai",
    "Delhi", "Pune", "Coimbatore", "Kochi",
    "Kolkata", "Jaipur"
]

stores = []

for i in range(1, 11):
    stores.append({
        "store_id": i,
        "store_name": f"Store_{i}",
        "city": cities[i - 1]
    })

stores_df = pd.DataFrame(stores)

# -----------------------------
# 3. ORDERS DATASET
# -----------------------------

orders = []

start_date = datetime(2026, 1, 1)

for i in range(1, 501):
    order_date = start_date + timedelta(
        days=random.randint(0, 90)
    )

    orders.append({
        "order_id": i,
        "customer_id": (
            None if i % 37 == 0 else random.randint(1001, 1100)
        ),
        "store_id": random.randint(1, 10),
        "order_date": order_date.strftime("%Y-%m-%d")
    })

orders_df = pd.DataFrame(orders)

# Add intentional duplicate records
orders_df = pd.concat(
    [orders_df, orders_df.iloc[[9, 19, 29]]],
    ignore_index=True
)

# -----------------------------
# 4. ORDER ITEMS DATASET
# -----------------------------

order_items = []

for i in range(1, 1001):
    quantity = random.randint(1, 5)

    # Intentional invalid quantity
    if i % 97 == 0:
        quantity = -2

    order_items.append({
        "order_item_id": i,
        "order_id": random.randint(1, 500),
        "product_id": random.randint(1, 50),
        "quantity": quantity
    })

order_items_df = pd.DataFrame(order_items)

# -----------------------------
# SAVE ALL DATASETS
# -----------------------------

datasets = {
    "products.csv": products_df,
    "stores.csv": stores_df,
    "orders.csv": orders_df,
    "order_items.csv": order_items_df
}

for filename, dataframe in datasets.items():
    path = f"{BASE_PATH}/{filename}"
    dataframe.to_csv(path, index=False)
    print(f"Created: {filename} | Rows: {len(dataframe)}")

print("\nAll RetailFlow datasets generated successfully!")
```
