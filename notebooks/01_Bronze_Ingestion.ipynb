```python

from pyspark.sql.functions import current_timestamp

# Source location
BASE_PATH = "/Volumes/dbacademy/retailflow/raw_data/"

# Read CSV files
orders_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(BASE_PATH + "orders.csv")

order_items_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(BASE_PATH + "order_items.csv")

products_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(BASE_PATH + "products.csv")

stores_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(BASE_PATH + "stores.csv")

print("All CSV files read successfully!")
```

```python
orders_df = orders_df.withColumn(
    "ingestion_timestamp", current_timestamp()
)

order_items_df = order_items_df.withColumn(
    "ingestion_timestamp", current_timestamp()
)

products_df = products_df.withColumn(
    "ingestion_timestamp", current_timestamp()
)

stores_df = stores_df.withColumn(
    "ingestion_timestamp", current_timestamp()
)
```

```python
TABLES = {
    "bronze_orders": orders_df,
    "bronze_order_items": order_items_df,
    "bronze_products": products_df,
    "bronze_stores": stores_df
}

for table_name, dataframe in TABLES.items():

    dataframe.write \
        .format("delta") \
        .mode("overwrite") \
        .saveAsTable(f"dbacademy.retailflow.{table_name}")

    print(f"Created table: {table_name}")
```
